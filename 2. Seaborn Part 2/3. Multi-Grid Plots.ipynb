### Multi-Grid Plots -:
1. Facet Grid and Facet Plot
2. Pair Grid and Pair Plot
3. Joint Grid and Joint plot

```python
import seaborn as sns
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

### 1. Facet Grid And Facet Plot -:
A 2nd way to Plot facet plots.

```python
#1. figure level -> relplot -> displot -> catplot -> lmplot
# 1st Method
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
# 2nd Method
g = sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.violinplot,'sex','total_bill')
```

```python
# Hue Parameter -:
g = sns.FacetGrid(data=tips,col='day',row='time',hue='smoker')
g.map(sns.boxplot,'sex','total_bill')
```

### 2. Pair Grid and Pair Plot -:
1. Plotting Pairwise Relationship
2. Same vs Same --> Histogram  (Pair Plot)
3. Different vs Different --> Scatter   (Pair Plot)
4. Same vs Same --> KDE Plot  (Hue Parameter in pair plot)

```python
#1. Pair Plot -: It's Automatically detects numerical columns in data set then it's plots scatter plot pair by pair between numerical columns.
sns.pairplot(iris)
```

```python
#2. Hue Parameter -:
sns.pairplot(iris,hue='species')
```

```python
#3. Pair Grid -:
g = sns.PairGrid(data=iris,hue='species')
# g.map
g.map(sns.scatterplot)
```

```python
#4. map_diag -> map_offdiag
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
```

```python
#5. map_diag -> map_upper -> map_lower
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
#6. vars -:
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

### 3. Joint Plot and Joint Grid -:

```python
#1. Joint Plot -:
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist')
```

```python
#2. Hue Parameter -:
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist',hue='sex')
```

```python
#3. Joint Grid -:
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.violinplot)
```
