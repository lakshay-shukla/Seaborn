### Figure level function -> `catplot`

```python
import seaborn as sns
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

### 1. Categorical Scatter Plot -:

- Stripplot
- Swarmplot

#### A. Strip Plot -: Categorical vs numerical

```python
#1. Axis level function -:
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
#2. using catplot -: figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
#3. jitter parameter -:
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=0.2)
```

```python
#4. hue parameter -:
sns.catplot(data=tips, x='day',y='total_bill',kind='strip',jitter=0.2,hue='sex')
```

#### B. Swarm Plot -: Better than Strip but for small data set.

```python
#1. Axis level function -:
sns.swarmplot(data=tips, x='day',y='total_bill')
```

```python
#2. using catplot -: figure level function
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
#3. hue parameter -:
sns.swarmplot(data=tips, x='day',y='total_bill',hue='sex')
```

### 2. Categorical Distribution Plots -:

- Boxplot
- Violinplot

#### A. Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

image.png

```python
#1. Axis Level -:
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
#2. Using catplot --> Figure Level
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
#3. hue parameter -:
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
#4. single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

```python
# faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='box',row='time')
```

#### B. Violinplot = (Boxplot + KDEplot)

```python
#1. Axis Level -:
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
#2. Figure Level -:
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

```python
#3. hue parameter -:
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex')
```

```python
#4. split parameter -:
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

```python
#5. faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='violin',row='time')
```

### 3. Categorical Estimate Plot -> for central tendency -:

- Barplot
- Pointplot
- Countplot

#### A. Barplot -:
1. The line on the bar is called error bar and it's shows the mean of the bar.
2. You can change it from mean to other mathematical functions.

```python
#1. Axis Level -:
sns.barplot(data=tips, x='day', y='total_bill')
```

```python
#2. Figure Level -:
sns.catplot(data=tips,x='day',y='total_bill',kind='bar')
```

```python
#3. hue parameter -:
sns.catplot(data=tips,x='day',y='total_bill', hue='sex', kind='bar')
```

```python
#4. Default value change from mean to others -: estimator=np.min, np.sum etc
import numpy as np
sns.barplot(data=tips, x='sex', y='total_bill',hue='smoker',estimator=np.min)
```

```python
#5. To remove error bars -: errorbar=None
sns.barplot(data=tips, x='sex', y='total_bill',errorbar=None)
```

```python
#6. faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='bar',row='time')
```

#### B. Point Plot -:
1. It's join mid point of the error bars.
2. When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
#1. Axis Level -:
sns.pointplot(data=tips, x='sex', y='total_bill')
```

```python
#2. Figure Level -:
sns.catplot(data=tips, x='sex', y='total_bill', kind='point')
```

```python
#3. hue parameter -:
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker')
```

```python
#4. To remove error bars -: errorbar=None
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker', errorbar=None)
```

```python
#5. faceting using catplot
sns.catplot(data=tips, x='sex',y='total_bill',col='smoker',kind='point',row='time')
```

#### C. Count Plot -:
A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
#1. Axis Level -:
sns.countplot(data=tips,x='sex',)
```

```python
#2. Figure Level -:
sns.catplot(data=tips,x='sex', kind='count')
```

```python
#3. hue parameter -:
sns.countplot(data=tips,x='sex',hue='day')
```
