### Why Seaborn ?
1. Provides a layer of abstraction hence simpler to use.
2. Better Aesthetics.
3. More graphs included.

#### Types of Function ( Levels of Function) -:
1. Figure Level
2. Axis Level

#### Main Classification -:

1. Relational Plot
2. Distribution Plot
3. Categorical Plot
4. Regression Plot
5. Matrix Plot
6. Multiplots

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

###  Relational Plot -:
A. To see the statistical relation between 2 or more variables.
B. Bivariate Analysis

#### Plots under this section -:
1. Scatter Plot
2. Line Plot

```python
tips = sns.load_dataset('tips')
tips
```

#### 1. Scatter Plot -:

```python
#1. Axis level function -: It's shape is rectangular.
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

```python
#2. relplot -> figure level -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

```python
#3. Using Different parameters in seaborn -> To add more information. (size, hue,style).

#A. Hue -:
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex')
```

```python
#B. size -:

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',size='size')
```

```python
#C. style -:
# relplot -> figure level -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
```

#### 2. Line Plot -:
To Continuity in graph.

```python
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
#1. Axis level function -:
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
#2. Using relplot (Figure level Function) -:
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line')
```

```python
#3. Using different parameters -: (hue,style,size)

#A. Hue -:
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country')
```

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country')
```

```python
#B. style -:
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country', style='continent')
```

```python
#C. size -:
sns.relplot(kind='line', data=temp_df, x='year', y='lifeExp', hue='country', style='continent', size='continent')
```

#### facet plot -> figure level function -> work with relplot
1. It will not work with scatterplot and lineplot (Axis Level Function)
2. You can make multiple plots on a particular/single categorical column and rows also.

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='line', col='sex', row='day')
```

```python
# column wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```
