### Distribution Plots -:
1. Used for univariate analysis
2. Used to find out the distribution
3. Range of the observation
4. Central Tendency
5. Is the data bimodal?
6. Are there outliers?

#### Plots under distribution plot

1. Histogram (histplot)
2. Kernel Density Estimation Plot (kdeplot)
3. rugplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
tips = sns.load_dataset('tips')
```

### 1. Histogram (histplot) -:
 1. figure level -> displot
 2. axis level -> histplot -> kdeplot -> rugplot
 3. It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
 4. Discrete bins are automatically set for categorical variables

```python
#1. Plotting univariate histogram -:
#A. Figure Level
sns.histplot(data=tips, x='total_bill')
```

```python
#B. Axis Level
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
#2. Bins parameter -:
sns.displot(data=tips, x='total_bill', kind='hist',bins=20)
```

```python
#3. countplot -:
sns.displot(data=tips, x='day', kind='hist')
```

```python
#4. Hue parameter
sns.displot(data=tips, x='tip', kind='hist',hue='sex')
```

```python
#A. element -> step
sns.displot(data=tips, x='tip', kind='hist',hue='sex',element='step')
```

```python
#5. Faceting using col and row -> not work on histplot function
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step')
```

### 2. Kernel Density Estimation Plot (kdeplot) -:
1. Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate.

```python
#1. Plotting kdeplot -:
#A. Axis Level
sns.kdeplot(data=tips,x='total_bill')
```

```python
#B. Figure level
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
#2. hue -> fill
#A. hue
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex')
```

```python
#B. Fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
```

```python
#3. Height and Aspect -:
#B. height
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10)
```

```python
#B. aspect -:
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

### 3. Rugplot -:
1. Plot marginal distributions by drawing ticks along the x and y axes.
2. This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
3. A Rug Plot is used with a graph.

```python
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```
